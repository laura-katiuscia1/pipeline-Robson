```python
#erve para manipular, limpar, analisar e transformar grandes volumes de dados em formato de tabela.
import pandas as pd

tabelas = pd.read_csv("estagio.csv", encoding='latin1', sep=';')

print("Primeiras linhas:")
print(tabelas.head())
```

```python
lendo = pd.read_csv("estagio.csv", encoding='latin1', sep=';')
print(f"Dados extraídos: {len(lendo)} linhas encontradas.")
```

```python
display(lendo.head())
```

```python
#media da idade das pessoas atendidas
idade_series = lendo['Nome, idade, cidade, observacao'].str.split(',').str[1]
media_idade = pd.to_numeric(idade_series, errors='coerce').mean()
print(media_idade)
```

```python
# 1. Separamos a coluna bagunçada em 4 novas colunas
# O nome da coluna no seu print parece ser "Nome, idade, cidade, observacao"
novo_df = lendo['Nome, idade, cidade, observacao'].str.split(',', expand=True)

# 2. Renomeamos as colunas para ficar organizado
novo_df.columns = ['nome', 'idade', 'cidade', 'observacao']

# 3. Importante: Converter a idade para número (pois o split cria textos)
novo_df['idade'] = pd.to_numeric(novo_df['idade'])

# 4. Agora sim, a média!
print(f"A média de idade é: {novo_df['idade'].mean()}")
```

```python
# Para ver a lista de todas as cidades cadastradas
print(novo_df['cidade'])

# Para ver quais cidades aparecem e quantas vezes (frequência)
print(novo_df['cidade'].value_counts())
```

```python
#Algritmo mostra  quantos moradores foram encontrados por cidade

from sklearn.preprocessing import LabelEncoder

# 1. Criamos o codificador
le = LabelEncoder()

# 2. Transformamos a coluna "cidade" em números (ex: Guarulhos = 0, São Paulo = 1)
novo_df['cidade_codigo'] = le.fit_transform(novo_df['cidade'])

# Agora, para o algoritmo de ML, você usaria 'cidade_codigo'
print(novo_df[['cidade', 'cidade_codigo']])
```

```python
import matplotlib.pyplot as plt

#
contagem_cidades = novo_df['cidade'].value_counts()

contagem_cidades.plot(kind='bar', color='skyblue')
plt.title('Quantidade de Moradores por Cidade')
plt.xlabel('Cidade')
plt.ylabel('Total de Pessoas')
plt.show()
```

```python
# contiuação do algoritmo
from sklearn.tree import DecisionTreeClassifier

# Definindo X_treino (features) e y_treino (variável alvo)
# Usaremos 'idade' e 'cidade_codigo' como features
X_treino = novo_df[['idade', 'cidade_codigo']]
# Para fins de exemplo, vamos usar 'cidade_codigo' como alvo.
# Você pode substituir isso pela sua variável alvo real.
y_treino = novo_df['cidade_codigo']

modelo = DecisionTreeClassifier()
modelo.fit(X_treino, y_treino)
print("Modelo treinado com sucesso!")
```

```python
#Histograma mostra faixa etária de idade das pessoas atendidas
plt.hist(novo_df['idade'], bins=5, color='salmon', edgecolor='black')
plt.title('Distribuição de Idade dos Moradores')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```

```python
#Matplotlib Gráfico de pizza

novo_df['cidade'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Percentual por Cidade')
plt.ylabel('') # Remove o nome da coluna lateral
plt.show()
```
