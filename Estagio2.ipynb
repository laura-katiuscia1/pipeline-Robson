```python
lendo = pd.read_csv("estagio1.csv", encoding='latin1', sep=';')
print(f"Dados extraídos: {len(lendo)} linhas encontradas.")
```

```python
display(lendo.head())
```

```python
# Média da idade das pessoas atendidas
# Corrigindo o nome da coluna para o nome exato presente no arquivo ('Nome, idade, cidade, observação,sexo')
idade_series = lendo['Nome, idade, cidade, observação,sexo'].str.split(',').str[1]
media_idade = pd.to_numeric(idade_series, errors='coerce').mean()
print(f"A média de idade é: {media_idade:.2f}")
```

```python
# 1. Separamos a coluna bagunçada em 5 novas colunas (incluindo o sexo)
# Corrigido: 'observação' com acento para corresponder ao arquivo 'estagio1.csv'
novo_df = lendo['Nome, idade, cidade, observação,sexo'].str.split(',', expand=True)

# 2. Renomeamos as colunas para ficar organizado (o arquivo original tem 5 valores separados por vírgula)
# Ajustando para receber as 5 colunas criadas pelo split
novo_df.columns = ['nome', 'idade', 'cidade', 'observacao', 'sexo']

# 3. Importante: Converter a idade para número (pois o split cria textos)
novo_df['idade'] = pd.to_numeric(novo_df['idade'], errors='coerce')

# 4. Agora sim, a média!
print(f"A média de idade é: {novo_df['idade'].mean():.2f}")
```

```python
# Para ver a lista de todas as cidades cadastradas
print(novo_df['cidade'])

# Para ver quais cidades aparecem e quantas vezes (frequência)
print(novo_df['cidade'].value_counts())
```

```python
#Algritmo mostra  quantos moradores foram encontrados por cidade

from sklearn.preprocessing import LabelEncoder

# 1. Criamos o codificador
le = LabelEncoder()

# 2. Transformamos a coluna "cidade" em números (ex: Guarulhos = 0, São Paulo = 1)
novo_df['cidade_codigo'] = le.fit_transform(novo_df['cidade'])

# Agora, para o algoritmo de ML, você usaria 'cidade_codigo'
print(novo_df[['cidade', 'cidade_codigo']])
```

```python
import matplotlib.pyplot as plt

#
contagem_cidades = novo_df['cidade'].value_counts()

contagem_cidades.plot(kind='bar', color='skyblue')
plt.title('Quantidade de Moradores por Cidade')
plt.xlabel('Cidade')
plt.ylabel('Total de Pessoas')
plt.show()
```

```python
# contiuação do algoritmo
from sklearn.tree import DecisionTreeClassifier

# Definindo X_treino (features) e y_treino (variável alvo)
# Usaremos 'idade' e 'cidade_codigo' como features
X_treino = novo_df[['idade', 'cidade_codigo']]
# Para fins de exemplo, vamos usar 'cidade_codigo' como alvo.
# Você pode substituir isso pela sua variável alvo real.
y_treino = novo_df['cidade_codigo']

modelo = DecisionTreeClassifier()
modelo.fit(X_treino, y_treino)
print("Modelo treinado com sucesso!")
```

```python
#Histograma mostra faixa etária de idade das pessoas atendidas
plt.hist(novo_df['idade'], bins=5, color='salmon', edgecolor='black')
plt.title('Distribuição de Idade dos Moradores')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.show()
```

```python
#Matplotlib Gráfico de pizza

novo_df['cidade'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Percentual por Cidade')
plt.ylabel('') # Remove o nome da coluna lateral
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Lendo o arquivo correto
lendo = pd.read_csv("estagio1.csv", encoding='latin1', sep=';')

# Ajustando o nome da coluna correto e dividindo em 5 partes (incluindo o sexo)
novo_df = lendo['Nome, idade, cidade, observação,sexo'].str.split(',', expand=True)
novo_df.columns = ['nome', 'idade', 'cidade', 'observacao', 'sexo']

# Removendo espaços extras que possam existir nos textos
novo_df['sexo'] = novo_df['sexo'].str.strip()

# Agora a coluna 'sexo' existe e o gráfico pode ser gerado!
if 'sexo' in novo_df.columns and not novo_df['sexo'].isna().all():
    novo_df['sexo'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'], edgecolor='black')
    plt.title('Quantidade de Pessoas por Sexo')
    plt.xlabel('Sexo')
    plt.ylabel('Quantidade')
    plt.xticks(rotation=0)
    plt.show()
else:
    print("Não foi possível gerar o gráfico. Colunas disponíveis:", list(novo_df.columns))
```
